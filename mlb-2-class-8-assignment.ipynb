```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = pd.read_csv('/kaggle/input/datasets/lovishbansal123/titanic-dataset/Titanic-Dataset.csv')
iris = pd.read_csv('/kaggle/input/datasets/himanshunakrani/iris-dataset/iris.csv')
tips = pd.read_csv('/kaggle/input/datasets/saurabhbadole/restaurant-tips-dataset/tips.csv')
```

# Visualize at least 1 graph of each category. You can take help from the reference notebook

# Univariate Analysis

# 1. Pie chart

```python
titanic['Sex'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Male vs Female Passengers')
plt.ylabel('')
plt.show()
```

# 2. Count plot

```python
sns.countplot(data = titanic,x='Pclass')
plt.title('Titanic Passenger Class Count Plot')
plt.show()
```

# 3. Kde + Rug plot

```python
plt.figure(figsize=(5, 3))
sns.kdeplot(data=iris, x='sepal_length')
sns.rugplot(data=iris, x='sepal_length')
plt.title('KDE + Rug Plot of Sepal Length')
plt.show()
```

# 4. Violin plot

```python
plt.figure(figsize=(7, 4))
sns.violinplot(y='total_bill', data=tips)
plt.title('Distribution of Total Bill')
plt.show()
```

# 5. Box plot

```python
plt.figure(figsize=(7, 4))
sns.boxplot(y='Age', data=titanic)
plt.title('Age Distribution')
plt.show()
```

# 6. Boxen plot

```python
plt.figure(figsize=(7, 4))
sns.boxenplot(y='tip', data=tips)
plt.title('Distribution of Tips')
plt.show()
```

# 7. Histogram

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=tips, x='total_bill', bins=20)
plt.title('Distribution of Total Bill')
plt.show()
```

# Bivariate Analysis

# 1. Scatter plot

```python
plt.figure(figsize=(9, 5))
sns.scatterplot(x='sepal_length', y='sepal_width', data=iris)
plt.title('Sepal Length vs Sepal Width')
plt.show()
```

# 2. Line plot

```python
avg_tip = tips.groupby('day')['tip'].mean()
plt.plot(avg_tip.index, avg_tip.values, marker='o')
plt.title('Average Tip by Day')
plt.xlabel('Day')
plt.ylabel('Average Tip')
plt.show()
```

# 3. Joint plot

```python
sns.jointplot(x='total_bill', y='tip', data=tips)
plt.show()
```

# 4. Hexbin plot

```python

plt.hexbin(tips['total_bill'], tips['tip'], gridsize=15)
plt.title('Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

# 5. 2D kde plot

```python
sns.kdeplot(data=iris, x='sepal_length',y='sepal_width')
plt.title('2D KDE Plot')
plt.show()
```

# 6. Box plot

```python
plt.figure(figsize=(7, 5))

sns.boxplot(x='Pclass', y='Age', data=titanic)

plt.title('Age Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Age')

plt.show()
```

# 7. Violin plot

```python

plt.figure(figsize=(8, 5))
sns.violinplot(data=tips, x="day", y="tip")
plt.title("Tip Distribution by Day")
plt.xlabel("Day")
plt.ylabel("Tip")
plt.show()
```

# 8. Swarm plot

# 9. Count plot

```python
sns.swarmplot(data=tips, x='day', y='tip')

plt.title('Tip by Day')
plt.show()
sns.swarmplot(data=tips, x='day', y='tip')

plt.title('Tip distribution by Day')
plt.show()
```

# Multivariate Analysis

# 1. Scatter plot

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x='sepal_length', y='petal_length',
                hue='species', data=iris)
plt.title('Sepal Length vs Petal Length by Species')
plt.show()
```

# 2. Pair plot

```python
sns.pairplot(iris, hue='species')
plt.show()
```

# 3. Heatmap

```python
plt.figure(figsize=(8, 6))
sns.heatmap(titanic.select_dtypes(include='number').corr(),
            annot=True,
            cmap='coolwarm')
plt.title('Titanic Correlation Heatmap')
plt.show()
```

# 4. FacetGrid plot

```python
g = sns.FacetGrid(tips, col="day", hue="sex", height=4, col_wrap=2)
g.map_dataframe(sns.scatterplot, x='total_bill', y='tip')
g.set_axis_labels('Total Bill', 'Tip')
plt.show()
```

# 5. PairGrid

```python

g = sns.PairGrid(
    tips,
    vars=["total_bill", "tip", "size"],
    hue="sex"
)
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
g.add_legend()
plt.show()
```

# 6. Joint plot

```python

sns.jointplot(data=tips, x='total_bill',y='tip',hue='sex')
plt.show()
```

# 7. Boxen plot

```python
sns.jointplot(data=tips,x='total_bill',  y='tip', hue='sex')
plt.show()
```
